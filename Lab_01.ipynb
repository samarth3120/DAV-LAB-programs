NumPy and Pandas basics array operations, Data Frame creation, indexing, slicing.

Use a Python 3 kernel. If packages are missing, run `%pip install numpy pandas` in that kernel.

```python
import numpy as np
import pandas as pd

# --- NumPy array operations ---
arr = np.array([10, 20, 30, 40, 50])
print('Array:', arr)
print('Mean:', arr.mean())
print('Reshaped 2D:\n', arr.reshape(5,1))
print('Sliced (index 1 to 3):', arr[1:4])

# --- Create a DataFrame ---
data = {'Name': ['Asha','Ravi','Meera','Karan'],
        'Marks': [88, 72, 91, 65],
        'Branch': ['CSE','ECE','CSE','ISE']}
df = pd.DataFrame(data)
print(df)

# --- Indexing and slicing ---
print('\nFirst 2 rows:\n', df.iloc[0:2])
print('\nMarks column:\n', df['Marks'])
print('\nRows where Marks > 70:\n', df[df['Marks'] > 70])
```
