# Lab 4 · Data Preprocessing

Handling missing data using imputation (mean, median, forward fill), and encoding categorical variables.

[Download the dataset from Kaggle](https://www.kaggle.com/datasets/dansbecker/melbourne-housing-snapshot) and save its CSV as `melb_data.csv`.

Place the CSV in the notebook's working directory.

Use a Python 3 kernel. If packages are missing, run `%pip install numpy pandas scikit-learn` in that kernel.

```python
# Import required libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

# Load the Melbourne Housing dataset
df = pd.read_csv("melb_data.csv")

# Display the first five rows
print("Original Dataset:")
print(df.head())

print("\nShape of dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
print(df.info())

print("\nStatistical summary:")
print(df.describe())

# Check for missing values
print("\nMissing Values Before Preprocessing:")
print(df.isnull().sum())

# ------------------------------------------------------------
# Mean Imputation
# ------------------------------------------------------------

# Create a copy of the dataset
df_mean = df.copy()

# Calculate the mean of BuildingArea
mean_building_area = df_mean["BuildingArea"].mean()

# Replace missing BuildingArea values with the mean
df_mean["BuildingArea"] = df_mean["BuildingArea"].fillna(
    mean_building_area
)

print("\nMean Building Area:", mean_building_area)

print("Missing BuildingArea values after Mean Imputation:")
print(df_mean["BuildingArea"].isnull().sum())

# ------------------------------------------------------------
# Median Imputation
# ------------------------------------------------------------

# Create a copy of the dataset
df_median = df.copy()

# Calculate the median of YearBuilt
median_year_built = df_median["YearBuilt"].median()

# Replace missing YearBuilt values with the median
df_median["YearBuilt"] = df_median["YearBuilt"].fillna(
    median_year_built
)

print("\nMedian Year Built:", median_year_built)

print("Missing YearBuilt values after Median Imputation:")
print(df_median["YearBuilt"].isnull().sum())

# ------------------------------------------------------------
# Forward Fill Imputation
# ------------------------------------------------------------

# Create a copy of the dataset
df_ffill = df.copy()

# Fill missing Car values using the previous value
df_ffill["Car"] = df_ffill["Car"].ffill()

print("\nMissing Car values after Forward Fill:")
print(df_ffill["Car"].isnull().sum())


# ------------------------------------------------------------
# Comparison of Imputation Methods
# ------------------------------------------------------------

comparison = pd.DataFrame({
    "Original BuildingArea": df["BuildingArea"].head(15),
    "Mean Imputed": df_mean["BuildingArea"].head(15),

    "Original YearBuilt": df["YearBuilt"].head(15),
    "Median Imputed": df_median["YearBuilt"].head(15),

    "Original Car": df["Car"].head(15),
    "Forward Filled": df_ffill["Car"].head(15)
})

print("\nComparison of Imputation Methods:")
print(comparison)

# ------------------------------------------------------------
# Handling Missing Categorical Values
# ------------------------------------------------------------

# Create a copy of the dataset
df_processed = df.copy()

# Find the most frequent value of CouncilArea
council_mode = df_processed["CouncilArea"].mode()[0]

# Replace missing CouncilArea values with the mode
df_processed["CouncilArea"] = df_processed["CouncilArea"].fillna(
    council_mode
)

print("\nCouncilArea Mode:", council_mode)

print("\nMissing Categorical Values After Imputation:")
print(df_processed[["CouncilArea"]].isnull().sum())

# ------------------------------------------------------------
# One-Hot Encoding
# ------------------------------------------------------------

# Convert Type and Method into numerical columns
df_encoded = pd.get_dummies(
    df_processed,
    columns=["Type", "Method"],
    dtype=int
)

print("\nDataset After One-Hot Encoding:")
print(df_encoded.head())

# ------------------------------------------------------------
# Label Encoding
# ------------------------------------------------------------

# Create a copy of the dataset
df_label = df.copy()

# Create LabelEncoder
label_encoder = LabelEncoder()

# Encode the CouncilArea column
df_label["CouncilArea_Encoded"] = label_encoder.fit_transform(
    df_label["CouncilArea"]
)

print("\nLabel Encoding of CouncilArea:")
print(
    df_label[
        ["CouncilArea", "CouncilArea_Encoded"]
    ].head(10)
)
```
