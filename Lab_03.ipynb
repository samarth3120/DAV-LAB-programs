# Lab 3 · Data Visualization using Matplotlib and Seaborn

histograms, box plots, scatter plots, pair plots, and correlation heatmaps.

[Download the dataset from Kaggle](https://www.kaggle.com/datasets/uciml/iris) and save its CSV as `Iris.csv`.

Place the CSV in the notebook's working directory.

Use a Python 3 kernel. If packages are missing, run `%pip install pandas matplotlib seaborn` in that kernel.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# LOAD THE DATASET

df = pd.read_csv("Iris.csv")

# VIEW THE FIRST 5 ROWS
print("First 5 rows of the dataset:")
print(df.head())

#  VIEW THE LAST 5 ROWS
print("\nLast 5 rows of the dataset:")
print(df.tail())

# CHECK THE SIZE OF THE DATASET
print("\nShape of the dataset:")
print(df.shape)

# VIEW COLUMN NAMES
print("\nColumn names:")
print(df.columns)

# CHECK DATASET INFORMATION
print("\nDataset information:")
df.info()
# Check missing values
print(df.isnull().sum())

# Remove missing values
df = df.dropna()

# Check missing values again
print(df.isnull().sum())

# Check duplicate rows
print(df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Check duplicates again
print(df.duplicated().sum())

# Display statistical summary
print(df.describe())

# Set graph style
sns.set_style("whitegrid")
# ---------------- HISTOGRAM ----------------

# Create graph area
plt.figure(figsize=(8, 5))

# Create histogram
sns.histplot(data=df, x="SepalLengthCm", bins=10)

# Add title
plt.title("Distribution of Sepal Length")

# Add X-axis label
plt.xlabel("Sepal Length (cm)")

# Add Y-axis label
plt.ylabel("Number of Flowers")

# Display graph
plt.show()
# ---------------- BOX PLOT ----------------

# Create graph area
plt.figure(figsize=(8, 5))

# Create box plot
sns.boxplot(data=df, x="Species", y="SepalLengthCm")

# Add title
plt.title("Sepal Length by Species")

# Add X-axis label
plt.xlabel("Species")

# Add Y-axis label
plt.ylabel("Sepal Length (cm)")

# Display graph
plt.show()

# ---------------- SCATTER PLOT ----------------
# Create graph area
plt.figure(figsize=(8, 5))

# Create scatter plot
sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="PetalLengthCm",
    hue="Species"
)
# Add title
plt.title("Sepal Length vs Petal Length")

# Add X-axis label
plt.xlabel("Sepal Length (cm)")

# Add Y-axis label
plt.ylabel("Petal Length (cm)")

# Display graph
plt.show()
# ---------------- PAIR PLOT ----------------

# Create pair plot
sns.pairplot(df, hue="Species")

# Display graph
plt.show()

# ---------------- CORRELATION ----------------

# Select numerical columns
numeric_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

# Calculate correlation
correlation = df[numeric_columns].corr()

# Display correlation matrix
print(correlation)

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```
